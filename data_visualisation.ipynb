```python
import pandas as pd
import numpy as np
import plotly.express as px

# Load the dataset
df = pd.read_excel(r"C:\Users\thota\OneDrive\Documents\2 - 1\Data Analytics and Visualization\Data visualisation\cleaned_ecommerce_dataset.xlsx")

# Display the first 5 rows
df.head()
```

```python
# Display the number of rows and columns
print("Dataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display dataset information
df.info()

# Statistical summary
df.describe()
```

```python
# 1.Bar Chart — Total Sales by Product Category

sales_by_category = df.groupby(
    "Product_Category",
    as_index=False
)["Total_Amount"].sum()

fig = px.bar(
    sales_by_category,
    x="Product_Category",
    y="Total_Amount",
    title="Total Sales by Product Category",
    labels={
        "Product_Category": "Product Category",
        "Total_Amount": "Total Sales"
    },
    text_auto=".2s"
)

fig.show()
```

```python
# 2.Pie Chart — Sales Distribution by Product Category


sales_by_category = df.groupby(
    "Product_Category",
    as_index=False
)["Total_Amount"].sum()

fig = px.pie(
    sales_by_category,
    names="Product_Category",
    values="Total_Amount",
    title="Sales Distribution by Product Category",
    hole=0.3
)

fig.show()
```

```python
# 3.Histogram — Customer Age Distribution

fig = px.histogram(
    df,
    x="Age",
    nbins=20,
    title="Customer Age Distribution",
    labels={
        "Age": "Customer Age"
    }
)

fig.show()
```

```python
# 4.Scatter Plot — Unit Price vs Total Amount

fig = px.scatter(
    df,
    x="Unit_Price",
    y="Total_Amount",
    color="Product_Category",
    title="Unit Price vs Total Order Amount",
    labels={
        "Unit_Price": "Unit Price",
        "Total_Amount": "Total Amount"
    },
    hover_data=["Quantity", "Product_Category"]
)

fig.show()
```

```python
# 5.Correlation Heatmap — Numerical Variables

numeric_df = df.select_dtypes(
    include=["number"]
)

correlation = numeric_df.corr()

fig = px.imshow(
    correlation,
    text_auto=".2f",
    aspect="auto",
    title="Correlation Heatmap of Numerical Variables"
)

fig.show()
```

```python
# 6.Line Chart — Monthly Sales Trend

df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.to_period(
    "M"
).astype(str)

monthly_sales = df.groupby(
    "Month",
    as_index=False
)["Total_Amount"].sum()

fig = px.line(
    monthly_sales,
    x="Month",
    y="Total_Amount",
    markers=True,
    title="Monthly Sales Trend",
    labels={
        "Month": "Month",
        "Total_Amount": "Total Sales"
    }
)

fig.show()
```

```python
# 7.Bar Chart — Number of Orders by Payment Method

payment_counts = df["Payment_Method"].value_counts(
).reset_index()

payment_counts.columns = [
    "Payment_Method",
    "Number_of_Orders"
]

fig = px.bar(
    payment_counts,
    x="Payment_Method",
    y="Number_of_Orders",
    title="Number of Orders by Payment Method",
    labels={
        "Payment_Method": "Payment Method",
        "Number_of_Orders": "Number of Orders"
    },
    text_auto=True
)

fig.show()
```

```python
# 8.Pie Chart — Returning vs New Customers

customer_counts = df[
    "Is_Returning_Customer"
].value_counts().reset_index()

customer_counts.columns = [
    "Customer_Type",
    "Count"
]

customer_counts["Customer_Type"] = customer_counts[
    "Customer_Type"
].map({
    True: "Returning Customer",
    False: "Non-Returning Customer"
})

fig = px.pie(
    customer_counts,
    names="Customer_Type",
    values="Count",
    title="Returning vs Non-Returning Customer Records"
)

fig.show()
```

```python
# 9.Bar Chart — Average Customer Rating by Product Category

rating_by_category = df.groupby(
    "Product_Category",
    as_index=False
)["Customer_Rating"].mean()

rating_by_category["Customer_Rating"] = (
    rating_by_category["Customer_Rating"].round(2)
)

fig = px.bar(
    rating_by_category,
    x="Product_Category",
    y="Customer_Rating",
    title="Average Customer Rating by Product Category",
    labels={
        "Product_Category": "Product Category",
        "Customer_Rating": "Average Rating"
    },
    text_auto=True
)

fig.show()
```

```python
# 10.Histogram — Delivery Time Distribution

fig = px.histogram(
    df,
    x="Delivery_Time_Days",
    nbins=15,
    title="Delivery Time Distribution",
    labels={
        "Delivery_Time_Days": "Delivery Time (Days)"
    }
)

fig.show()
```

```python
# 11.Box Plot — Order Amount by Product Category

fig = px.box(
    df,
    x="Product_Category",
    y="Total_Amount",
    color="Product_Category",
    title="Order Amount Distribution by Product Category",
    labels={
        "Product_Category": "Product Category",
        "Total_Amount": "Total Order Amount"
    }
)

fig.show()
```

```python
# 12.Bar Chart — Top 10 Cities by Sales

sales_by_city = df.groupby(
    "City",
    as_index=False
)["Total_Amount"].sum()

top_10_cities = sales_by_city.nlargest(
    10,
    "Total_Amount"
)

fig = px.bar(
    top_10_cities,
    x="City",
    y="Total_Amount",
    title="Top 10 Cities by Total Sales",
    labels={
        "City": "City",
        "Total_Amount": "Total Sales"
    },
    text_auto=".2s"
)

fig.show()
```

```python
# 13.Bar Chart — Orders by Device Type

device_counts = df["Device_Type"].value_counts(
).reset_index()

device_counts.columns = [
    "Device_Type",
    "Number_of_Orders"
]

fig = px.bar(
    device_counts,
    x="Device_Type",
    y="Number_of_Orders",
    title="Number of Orders by Device Type",
    labels={
        "Device_Type": "Device Type",
        "Number_of_Orders": "Number of Orders"
    },
    text_auto=True
)

fig.show()
```

```python
# 14.Scatter Plot — Session Duration vs Pages Viewed

fig = px.scatter(
    df,
    x="Session_Duration_Minutes",
    y="Pages_Viewed",
    color="Is_Returning_Customer",
    title="Session Duration vs Pages Viewed",
    labels={
        "Session_Duration_Minutes": "Session Duration (Minutes)",
        "Pages_Viewed": "Pages Viewed",
        "Is_Returning_Customer": "Returning Customer"
    }
)

fig.show()
```

```python
# 15.Bar Chart — Average Order Amount by Gender

amount_by_gender = df.groupby(
    "Gender",
    as_index=False
)["Total_Amount"].mean()

amount_by_gender["Total_Amount"] = (
    amount_by_gender["Total_Amount"].round(2)
)

fig = px.bar(
    amount_by_gender,
    x="Gender",
    y="Total_Amount",
    title="Average Order Amount by Gender",
    labels={
        "Gender": "Gender",
        "Total_Amount": "Average Order Amount"
    },
    text_auto=True
)

fig.show()
```

```python
# 16.Geographical Map — Sales by City

import pandas as pd
import plotly.express as px
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

# Load the cleaned dataset
df = pd.read_excel(
    r"C:\Users\thota\OneDrive\Documents\2 - 1\Data Analytics and Visualization\Data visualisation\cleaned_ecommerce_dataset.xlsx"
)

# Calculate total sales for each city
city_sales = df.groupby("City", as_index=False)["Total_Amount"].sum()

# Set up geocoding
geolocator = Nominatim(user_agent="ecommerce_sales_visualization")
geocode = RateLimiter(
    geolocator.geocode,
    min_delay_seconds=1,
    max_retries=2
)

# Get geographical coordinates for each city
city_sales["Location"] = city_sales["City"].apply(
    lambda city: geocode(f"{city}, India")
)

city_sales["Latitude"] = city_sales["Location"].apply(
    lambda loc: loc.latitude if loc else None
)

city_sales["Longitude"] = city_sales["Location"].apply(
    lambda loc: loc.longitude if loc else None
)

# Remove cities whose coordinates could not be found
city_sales = city_sales.dropna(
    subset=["Latitude", "Longitude"]
)

# Create the world map
fig = px.scatter_geo(
    city_sales,
    lat="Latitude",
    lon="Longitude",
    size="Total_Amount",
    color="Total_Amount",
    hover_name="City",
    hover_data={
        "Total_Amount": ":,.2f",
        "Latitude": False,
        "Longitude": False
    },
    color_continuous_scale="YlOrRd",
    projection="natural earth",
    title="🌍 Geographical Distribution of E-commerce Sales by City",
    size_max=30
)

fig.update_layout(
    geo=dict(
        showland=True,
        landcolor="lightgray",
        showocean=True,
        oceancolor="lightblue",
        showcountries=True,
        countrycolor="white",
        showcoastlines=True
    ),
    height=650
)

fig.show()
```
