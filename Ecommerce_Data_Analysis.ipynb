```python
# 1. Total Revenue
total_revenue = df["Total_Amount"].sum()

# 2. Total Orders
total_orders = df["Order_ID"].nunique()

# 3. Total Quantity Sold
total_quantity = df["Quantity"].sum()

# 4. Average Order Value
average_order_value = total_revenue / total_orders if total_orders else 0

# 5. Average Selling Price per Unit
average_selling_price = (
    total_revenue / total_quantity if total_quantity else 0
)

# 6. Total Unique Customers
total_customers = df["Customer_ID"].nunique()

# 7. Returning Customers
returning_customers = df.loc[
    df["Is_Returning_Customer"], "Customer_ID"
].nunique()

# 8. Average Discount Amount
average_discount = df["Discount_Amount"].mean()

print("========== BUSINESS KPIs ==========")
print(f"Total Revenue: {total_revenue:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Average Order Value: {average_order_value:,.2f}")
print(f"Average Selling Price per Unit: {average_selling_price:,.2f}")
print(f"Total Unique Customers: {total_customers:,}")
print(f"Returning Customers: {returning_customers:,}")
print(f"Average Discount Amount: {average_discount:,.2f}")
```

```python
# -------------------Sales Trend Analysis--------------------
```

```python
# Extract year and month
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

# Create a year-month period
df["Year_Month"] = df["Date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("Year_Month").agg(
    Total_Revenue=("Total_Amount", "sum"),
    Total_Orders=("Order_ID", "nunique"),
    Total_Quantity=("Quantity", "sum"),
    Average_Order_Amount=("Total_Amount", "mean")
)

display(monthly_sales.round(2))
```

```python
highest_month = monthly_sales["Total_Revenue"].idxmax()
lowest_month = monthly_sales["Total_Revenue"].idxmin()

print("Highest Revenue Month:", highest_month)
print(
    "Revenue:",
    round(monthly_sales.loc[highest_month, "Total_Revenue"], 2)
)

print("\nLowest Revenue Month:", lowest_month)
print(
    "Revenue:",
    round(monthly_sales.loc[lowest_month, "Total_Revenue"], 2)
)
```

```python
monthly_sales["Revenue_Growth_Percent"] = (
    monthly_sales["Total_Revenue"].pct_change() * 100
)

display(monthly_sales.round(2))
```

```python
# Calculate revenue for each unique order
order_revenue = df.groupby("Order_ID")["Total_Amount"].sum()

# Calculate monthly revenue for each order
order_month = df.groupby("Order_ID")["Date"].min().dt.to_period("M")

monthly_order_data = pd.DataFrame({
    "Order_Revenue": order_revenue,
    "Year_Month": order_month
})

# Calculate monthly AOV
monthly_aov = monthly_order_data.groupby("Year_Month").agg(
    Total_Revenue=("Order_Revenue", "sum"),
    Total_Orders=("Order_Revenue", "count"),
    Average_Order_Value=("Order_Revenue", "mean")
)

display(monthly_aov.round(2))
```

```python
# --------------------------Statistical Analysis-------------------------
```

```python
# Select important numerical columns
numeric_columns = [
    "Age",
    "Unit_Price",
    "Quantity",
    "Discount_Amount",
    "Total_Amount",
    "Session_Duration_Minutes",
    "Pages_Viewed",
    "Delivery_Time_Days",
    "Customer_Rating"
]

# Generate statistical summary
statistical_summary = df[numeric_columns].describe().T

# Add median and mode
statistical_summary["Median"] = df[numeric_columns].median()
statistical_summary["Mode"] = df[numeric_columns].mode().iloc[0]

display(statistical_summary.round(2))
```

```python
correlation_columns = [
    "Unit_Price",
    "Quantity",
    "Discount_Amount",
    "Total_Amount",
    "Session_Duration_Minutes",
    "Pages_Viewed",
    "Delivery_Time_Days",
    "Customer_Rating"
]

correlation_matrix = df[correlation_columns].corr()

display(correlation_matrix.round(3))
```

```python
outlier_columns = [
    "Unit_Price",
    "Quantity",
    "Discount_Amount",
    "Total_Amount",
    "Delivery_Time_Days"
]

outlier_results = []

for column in outlier_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append({
        "Column": column,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": len(outliers) / len(df) * 100
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

# 10. Final Business Insights and Recommendations

## A. Key Business Insights

1. **Overall Revenue:** The dataset records approximately 4.92 million in total revenue across 5,000 orders.

2. **Top-Performing Category:** Electronics generates the highest revenue, approximately 2.33 million, making it the strongest category by revenue.

3. **Product Demand:** Sports records the highest quantity sold among the main product categories, with 1,518 units.

4. **Monthly Sales Performance:** January 2024 records the highest monthly revenue in the available period, at approximately 396,874.92.

5. **Geographical Performance:** Istanbul generates the highest city-level revenue, approximately 1.33 million, contributing around 27.14% of total revenue.

6. **Customer Behaviour:** Approximately 59.8% of customers are marked as returning customers in the dataset. However, each customer appears only once, so repeat purchasing cannot be independently verified.

7. **Variable Relationships:** Unit Price has a strong positive correlation with Total Amount (0.791). Quantity also has a moderate positive correlation with Total Amount (0.329).

8. **Potential Outliers:** Discount Amount has the highest number of potential outliers, followed by Total Amount and Unit Price. These records should be investigated before any changes are made.

## B. Business Recommendations

1. **Strengthen Electronics Sales:** Maintain the availability of popular electronics products and evaluate suitable promotional strategies to support revenue growth.

2. **Promote Other Categories:** Investigate ways to improve sales in lower-revenue categories such as Books, Food, and Beauty.

3. **Improve Regional Strategies:** Study Istanbul's successful sales performance and evaluate opportunities to improve sales in lower-revenue cities.

4. **Review Discount Policies:** Examine unusually large discounts to ensure that promotional activities follow business rules.

5. **Monitor Delivery Performance:** Investigate unusually long delivery times and identify opportunities to improve fulfilment.

6. **Validate High-Value Orders:** Review unusually large order amounts and unit prices to distinguish legitimate transactions from possible data errors.

7. **Improve Customer Data Tracking:** Record multiple transactions for each customer where applicable. This would support more reliable repeat-purchase, customer-retention, and lifetime-value analysis.

## C. Limitations

* The dataset does not contain cost or profit information, so actual profit and profit margin cannot be calculated.
* Each customer appears only once, limiting independent verification of repeat purchasing.
* The currency is not specified in the dataset.
* The IQR method identifies potential outliers but does not prove that they are errors.
* Correlation indicates association and does not establish causation.

## D. Conclusion

Exploratory and statistical analysis of the cleaned e-commerce dataset revealed useful insights into revenue, product category performance, monthly sales, customer characteristics, geographical performance, variable relationships, and potential outliers.

Electronics is the leading category by revenue, Istanbul is the leading city by revenue, and January 2024 is the highest-revenue month in the analysed period. Correlation analysis indicates that unit price is strongly associated with total order amount, while outlier detection highlights records that may require further investigation.

These findings can help businesses make more informed decisions about product sales, promotions, regional performance, and data quality. Additional cost and customer transaction history data would enable deeper profitability and customer-retention analysis.
